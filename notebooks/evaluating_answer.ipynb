```python
import pandas as pd 
```

```python
df_before = pd.read_csv("/Users/bern/Documents/GitHub/fyp-chatbot/tests/test_results/1702_step1.csv")
df_before
```

```python
df_after = pd.read_csv("/Users/bern/Documents/GitHub/fyp-chatbot/tests/test_results/1702_step3_50ans_first200q.csv")
df_after
```

```python
merged_df = pd.merge(df_before, df_after, on=['conversation_id','message_order', 'query'], how='inner', suffixes=('_before', '_after'))
merged_df
```

```python
# remove unused columns
merged_df = merged_df.drop(columns=['cost_before', 'cost_after', 'tokens_before', 'tokens_after'])
merged_df
```

```python
filtered_df = merged_df[merged_df['chatbot_response_before'] != merged_df['chatbot_response_after']]
filtered_df
```

```python
# export to xlsx
filtered_df.to_excel("/Users/bern/Documents/GitHub/fyp-chatbot/tests/test_results/1702_step1_step3_50ans_first200q.xlsx", index=False)
```
