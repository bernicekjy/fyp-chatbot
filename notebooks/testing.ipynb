## Test with chatlog data

### Test chatbot function

```python
print("Number of conversations: ", len(chatlog["chatlog"]))
```

```python
# cut down to sample chatlog
sample_chatlog = chatlog["chatlog"][108]
sample_chatlog
```

```python
results = []
for conversation in sample_chatlog:
    id = conversation["_id"]['$oid']

    message_counter = 1

    # printing response and metadata for one conversation
    for message in conversation["messages"]:

        # generate chatbot response for each user message
        if message["role"]=="user":
            # output user's query
            query = message["content"]
            result = bot.answer_this(query=query)
            
            result.update({"query":query, "message_order": message_counter, "conversation_id":id})

            # update message counter
            message_counter+=1
            
            results.append(result)
    
    # clear chat history since conversation is over
    bot.clear_chat_history()


results
```

```python
# convert results to dataframe
df = pd.DataFrame(data=results)

# reorder columns
ordered_columns = ['conversation_id', 'message_order', 'query','chatbot_response', 'context', 'cost', 'tokens' ]
df = df[ordered_columns]

df
```

```python
# export to csv

test_results_dir = "test_results"

df.to_csv(path_or_buf=test_results_dir+os.sep+"basic_indexing_results.csv", index=False)
```
